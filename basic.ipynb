```python
import pandas as pd

data = {
    "name": ["Maham", "Aleena", "Sara", "Ayesha", "Hina"],
    "age": [21, 22, 19, 24, 20],
    "marks": [85, 92, 78, 95, 88]
}

df = pd.DataFrame(data)  #dict to table

print(df)
print("Average marks:", df["marks"].mean())  #average of marks coloumn
print("Highest marks:", df["marks"].max())   #max marks in the mark coloumn
print("Lowest marks:", df["marks"].min())    #min marks in the mark coloumn

print("\nStudents scoring above 85:")
print(df[df["marks"] > 85]) #condition
```

```python

import pandas as pd

# Sample student data to explore
data = {
    "name": ["Maham", "Aleena", "Sara", "Ayesha", "Hina"],
    "age": [21, 22, 19, 24, 20],
    "marks": [85, 92, 78, 95, 88]
}

df = pd.DataFrame(data)

# Check how many rows and columns the DataFrame contains
print("Dataset dimensions:", df.shape)

# Display all column names
print("\nColumn names:")
print(df.columns)

# Check the data type of each column
print("\nColumn data types:")
print(df.dtypes)

# Display a summary of the DataFrame
# use whenever you load a new CSV or Excel dataset and want to check its columns, data types, missing values, and overall structure
print("\nDataset information:")   
df.info()   

# Get statistical information about numerical columns
print("\nStatistical summary:")
print(df.describe())
# df.describe(include="all") to include text coloumns

# Display the first and last five rows (by default)
print("\nFirst five rows:")    
print(df.head())
# print(df.head(2)) display first 2 rows

print("\nLast five rows:")     
print(df.tail())
# print(df.tail(2)) display last 2 rows
```

```python
import pandas as pd

data = {
    "name": ["Maham", "Aleena", "Sara", "Ayesha", "Hina"],
    "age": [21, 22, 19, 24, 20],
    "marks": [85, 92, 78, 95, 88]
}

df = pd.DataFrame(data)

# loc means select by index label
# This selects the row whose index label is 0
print(df.loc[0])

#To select a particular value
print(df.loc[0, "name"])

# iloc means select by integer position
# This selects the first row because Python positions start at zero
print(df.iloc[0])

# To retrieve a particular value
print(df.iloc[0, 1])
```

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Ali", "Sara", "Ahmed", "Hina"],
    "total": [80, 120, 95, 150]
})
```

```python
# Checks whether each value in the total column is greater than or equal to 100 and returns true or false
df["total"] >= 100
```

```python
# To keep the rows where the condition is True and excludes rows where it is False, it returns number
df[df["total"] >= 100]
```
