```python
import pandas as pd
```

```python
user = {                           
    "name" : ["Maham","Aleena"], 
    "age" : [21,22],
}
print(user.get("name"))
```

```python
df = pd.DataFrame(user) #Convert the dictionary into a table
print(df)
```

```python
df["name"] #Use the DataFrame name followed by the column name in square brackets to display name only
```

```python
df["age"]   #df[["age","name"]] #use 2 square brackets if you want to retrieve more than one coloumn using indexing
```

```python
df["age"].mean() #Pandas has a built-in method for calculating an average.
```

```python
df[df['age'] > 21] #Use a comparison operator inside the DataFrame column selection.
```

```python
import pandas as pd

data = {
    "name": ["Maham", "Aleena", "Sara", "Ayesha", "Hina"],
    "age": [21, 22, 19, 24, 20],
    "marks": [85, 92, 78, 95, 88]
}

df = pd.DataFrame(data)  #dict to table

print(df)
print("Average marks:", df["marks"].mean())  #average of marks coloumn
print("Highest marks:", df["marks"].max())   #max marks in the mark coloumn
print("Lowest marks:", df["marks"].min())    #min marks in the mark coloumn

print("\nStudents scoring above 85:")
print(df[df["marks"] > 85]) #condition
```
